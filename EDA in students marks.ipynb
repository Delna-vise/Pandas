```python
subjects = [
    'Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics'
]

df[subjects] = df[subjects].fillna(df[subjects].mean())
df
```

```python
df = df.drop_duplicates()
df
```

```python
# Average mark
df[['Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics']].mean()

# Highest mark
df[['Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics']].max()

# Lowest mark
df[['Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics']].min()

# Median mark
df[['Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics']].median()

# Standard deviation
df[['Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics']].std()

# Sort students by marks
subjects = ['Maths', 'Physics', 'Chemistry', 'English',
            'Biology', 'Economics', 'History', 'Civics']

df['Average_Marks'] = df[subjects].mean(axis=1)
```

```python
subjects = [
    'Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics'
]

for col in subjects:
    plt.figure(figsize=(7, 4))

    plt.hist(df[col], bins=10, edgecolor='black')

    plt.title(f'Distribution of {col} Marks')
    plt.xlabel('Marks')
    plt.ylabel('Number of Students')

    plt.show()
```

```python
for col in subjects:
    plt.figure(figsize=(7, 4))

    sns.boxplot(x=df[col])

    plt.title(f'Box Plot of {col} Marks')
    plt.xlabel('Marks')

    plt.show()
```

```python
subject_avg = df[subjects].mean()

for col in subjects:
    plt.figure(figsize=(6, 4))

    plt.bar(col, subject_avg[col])

    plt.title(f'Average Marks in {col}')
    plt.xlabel('Subject')
    plt.ylabel('Average Marks')

    plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.boxplot(data=df[subjects])

plt.title("Box Plot of Student Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(rotation=45)

plt.show()
```

```python
subject_avg = df[subjects].mean()

plt.figure(figsize=(10, 5))

subject_avg.plot(kind='bar')

plt.title("Average Marks by Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)

plt.show()
```

```python
sns.countplot(data=df, x='Gender')

plt.title("Number of Students by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Maths',
    y='Physics',
    hue='Gender'
)

plt.title("Maths vs Physics Marks")
plt.show()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    df[subjects].corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Between Subjects")

plt.show()
```

```python
for col in subjects:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) | (df[col] > upper)
    ]

    print(f"{col}:")
    print(outliers[['Name', col]])
```

```python
import pandas as pd

df = pd.read_csv(r"C:\Users\delna\Downloads\student_marks.csv")
df
```

```python
#1. Application: Probability using Z-score
```

```python
from scipy.stats import norm

mean = df['Maths'].mean()
std = df['Maths'].std()

mark = 60

zscore = (mark - mean) / std

print("Mean:", mean)
print("Standard Deviation:", std)
print("Z-score:", zscore)

probability = norm.cdf(zscore)

print("Probability of scoring below 60:", probability)
```

```python
#2. Application: Outlier Detection using Z-score
```

```python
from scipy.stats import zscore

threshold = 3

data = df['Maths']

z_scores = zscore(data)

outliers = df[np.abs(z_scores) > threshold]

print("Outliers:")
print(outliers[['Name', 'Maths']])
```

```python
#3. Application: Feature Scaling using StandardScaler
```

```python
from sklearn.preprocessing import StandardScaler

subjects = [
    'Maths', 'Physics', 'Chemistry', 'English',
    'Biology', 'Economics', 'History', 'Civics'
]

data = df[subjects]

scaler = StandardScaler()

# Fit the scaler to the data
scaler.fit(data)

# Display mean and variance
print("Mean:")
print(scaler.mean_)

print("Variance:")
print(scaler.var_)

# Transform the data
scaled_data = scaler.transform(data)

print("Scaled Data:")
print(scaled_data)
```

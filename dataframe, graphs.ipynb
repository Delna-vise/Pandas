```python
q1=df2["age"].quantile(q=0.25)
q3=df2["age"].quantile(q=0.75)
IQR=q3-q1
upper_bound=q3+1.5*IQR
lower_bound=q1-1.5*IQR
print(upper_bound)
print(lower_bound)
```
